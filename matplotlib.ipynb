## matplot lib example workflow

```python
%matplotlib inline
import matplotlib.pyplot as plt

#prepare data
x= [1,2,3,4]
y=[11,22,33,44]
#set up plot
fig, ax = plt.subplots(figsize=(2,2))
#plot data
ax.plot(x,y)
#customise plot
ax.set(title="simple plot",xlabel="x-axis",ylabel="y-label")
#save plot
fig.savefig("../ztmDS/zero-to-mastery-ml-master/zero-to-mastery-ml-master/images/simpleplot.jay.png")
#print plot
plt.show()
```

## histograms and subplots


```python
#from dictionaries
butter_prices = {"almond":10,"peanut":8,"cashew":12}
fig, ax = plt.subplots()
ax.set(title="butter store",xlabel="butter",ylabel="prices")
# ax.barh(butter_prices.keys(),butter_prices.values())
ax.barh(list(butter_prices.keys()),list(butter_prices.values()))

plt.show()
```

```python
#histogram
x= np.random.randn(1000)
fig, ax = plt.subplots()
ax.hist(x)
plt.show()
```

# subplots

```python
#option 1
fig , ((ax1,ax2),(ax3,ax4)) = plt.subplots(nrows=2,ncols=2,figsize=(10,5))

#plot to different axis
ax1.plot(x,x/2)
ax2.scatter(np.random.random(10),np.random.random(10))
ax3.bar(butter_prices.keys(),butter_prices.values())
ax4.hist(butter_prices.values())
plt.show()
```

```python
# option 2

fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,5))

#plot to index

ax[0,0].plot(x,x/2)
ax[0,1].scatter(np.random.random(10),np.random.random(10))
ax[1,0].bar(butter_prices.keys(),butter_prices.values())
ax[1,1].hist(butter_prices.values())
plt.show()
```

```python
heart_disease.head()
```

```python
over_50 = heart_disease[heart_disease["age"]>50]
over_50
```

```python
over_50.plot(kind="scatter",x="age",y="chol",c="target")
plt.show()
```

```python
fig , ax = plt.subplots(figsize=(10,6))
over_50.plot(kind="scatter",x="age",y="chol",c="target",ax=ax)
ax.set_xlim([45,100])
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,6))

scatter = ax.scatter(x=over_50["age"],y=over_50["chol"],c=over_50["target"])

ax.set(title="heart disease and cholestrol level", xlabel="age",ylabel="cholestrol")

ax.legend(*scatter.legend_elements(),title="target")


ax.axhline(over_50["chol"].mean(), linestyle="--")

plt.show()
```
